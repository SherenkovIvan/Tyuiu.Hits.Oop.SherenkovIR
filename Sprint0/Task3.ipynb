Описание задачи:

Создать базовый класс CreditCard в C#, который будет представлять информацию о кредитных картах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу CreditCard:

• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок действия (ExpiryDate).

• Методы:

o GetInfo(): метод для получения информации о кредитной карте в виде строки.

o Pay(): метод для оплаты покупки с использованием карты.

o CheckBalance(): метод для проверки баланса на карте.

Требования к производным классам:

1.GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть переопределен для добавления информации о получении бонусных миль при оплате покупки.

2.PremiumCard (PremiumCreditCard): Должен содержать дополнительные атрибуты, такие как Ассистент поддержки (SupportAssistant). Метод CheckBalance() должен быть переопределен для предоставления возможности обратиться за помощью к ассистенту поддержки в случае проблем с балансом.

3.CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Компания (Company). Метод GetInfo() должен быть переопределен для включения информации о компании в описании карты.



<h2 style="color:DodgerBlue">Реализация:</h2>


```
public interface ICreditCard
{
    string CardNumber { get; set; }
    string HolderName { get; set; }
    DateTime ExpiryDate { get; set; }
    
    string GetInfo();
    void Pay(decimal amount);
    void CheckBalance();
}

public interface IGoldCreditCard : ICreditCard
{
    int BonusMiles { get; set; }
}

public interface IPremiumCreditCard : ICreditCard
{
    string SupportAssistant { get; set; }
}

public interface ICorporateCreditCard : ICreditCard
{
    string Company { get; set; }
}

public class CreditCard : ICreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public DateTime ExpiryDate { get; set; }
    protected decimal _balance;

    public CreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal startBalance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        _balance = startBalance;
    }

    public virtual string GetInfo() => $"Карта: {CardNumber}, Владелец: {HolderName}, Срок: {ExpiryDate:MM/yy}";

    public virtual void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }
        
        if (amount > _balance)
        {
            Console.WriteLine($"Недостаточно средств! Баланс: {_balance:F2} руб.");
            return;
        }
        
        _balance -= amount;
        Console.WriteLine($"Оплачено: {amount:F2} руб. Остаток: {_balance:F2} руб.");
    }

    public virtual void CheckBalance() => Console.WriteLine($"Баланс карты {CardNumber}: {_balance:F2} руб.");
}

public class GoldCreditCard : CreditCard, IGoldCreditCard
{
    public int BonusMiles { get; set; }

    public GoldCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal startBalance, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, startBalance) => BonusMiles = bonusMiles;

    public override void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }
        
        if (amount > _balance)
        {
            Console.WriteLine($"Недостаточно средств! Баланс: {_balance:F2} руб.");
            return;
        }
        
        _balance -= amount;
        int miles = (int)(amount / 100);
        BonusMiles += miles;
        Console.WriteLine($"Оплачено: {amount:F2} руб. Начислено миль: {miles}. Всего миль: {BonusMiles}. Остаток: {_balance:F2} руб.");
    }

    public override string GetInfo() => base.GetInfo() + $", Мили: {BonusMiles}";
}

public class PremiumCreditCard : CreditCard, IPremiumCreditCard
{
    public string SupportAssistant { get; set; }

    public PremiumCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal startBalance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, startBalance) => SupportAssistant = supportAssistant;

    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"Персональный ассистент: {SupportAssistant}");
        if (_balance < 0)
        {
            Console.WriteLine("Внимание! Баланс отрицательный. Обратитесь к ассистенту за помощью!");
        }
        else
        {
            Console.WriteLine("Если у вас проблемы с балансом, обратитесь к ассистенту!");
        }
    }

    public override string GetInfo() => base.GetInfo() + $", Ассистент: {SupportAssistant}";
}

public class CorporateCreditCard : CreditCard, ICorporateCreditCard
{
    public string Company { get; set; }

    public CorporateCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal startBalance, string company)
        : base(cardNumber, holderName, expiryDate, startBalance) => Company = company;

    public override string GetInfo() => $"КОРПОРАТИВНАЯ КАРТА\nКомпания: {Company}\n{base.GetInfo()}";

    public override void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }
        
        if (amount > _balance)
        {
            Console.WriteLine($"Недостаточно корпоративных средств! Баланс: {_balance:F2} руб.");
            Console.WriteLine($"Уведомление отправлено в финансовый отдел {Company}");
            return;
        }
        
        _balance -= amount;
        Console.WriteLine($"Корпоративный платеж от {Company}: {amount:F2} руб. Остаток: {_balance:F2} руб.");
    }
}

// Тестирование
Console.WriteLine("=== ИНФОРМАЦИЯ О КАРТАХ ===\n");

CreditCard card = new CreditCard("123456", "Тимофей", new DateTime(2024, 11, 28), 47652);
Console.WriteLine("Обычная карта:");
card.GetInfo();
card.Pay(2300);
card.CheckBalance();
Console.WriteLine();

GoldCreditCard gold = new GoldCreditCard("678365", "Пелогея", new DateTime(2026, 12, 18),320000, 10);
Console.WriteLine("Золотая карта:");
gold.GetInfo();
gold.Pay(17000);
gold.Pay(2000);
gold.CheckBalance();
Console.WriteLine();

PremiumCreditCard premium = new PremiumCreditCard("123123", "Камилла", new DateTime(2026, 3, 31), -3000, "Елизаветта");
Console.WriteLine("Премиум карта:");
premium.GetInfo();
premium.CheckBalance();
Console.WriteLine();

CorporateCreditCard corporate = new CorporateCreditCard("777777", "Сергей", new DateTime(2024, 9, 30), 200000, "ООО Пупупу");
Console.WriteLine("Корпоративная карта:");
corporate.GetInfo();
corporate.Pay(50000);
corporate.CheckBalance();
Console.WriteLine();

Console.WriteLine("=== ИТОГОВАЯ ИНФОРМАЦИЯ ===");
card.GetInfo();
Console.WriteLine();
gold.GetInfo();
Console.WriteLine();
premium.GetInfo();
Console.WriteLine();
corporate.GetInfo();
```
