Описание задачи:

Создать базовый класс CreditCard в C#, который будет представлять информацию о кредитных картах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу CreditCard:

• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок действия (ExpiryDate).

• Методы:

o GetInfo(): метод для получения информации о кредитной карте в виде строки.

o Pay(): метод для оплаты покупки с использованием карты.

o CheckBalance(): метод для проверки баланса на карте.

Требования к производным классам:

1.GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть переопределен для добавления информации о получении бонусных миль при оплате покупки.

2.PremiumCard (PremiumCreditCard): Должен содержать дополнительные атрибуты, такие как Ассистент поддержки (SupportAssistant). Метод CheckBalance() должен быть переопределен для предоставления возможности обратиться за помощью к ассистенту поддержки в случае проблем с балансом.

3.CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Компания (Company). Метод GetInfo() должен быть переопределен для включения информации о компании в описании карты.



<h2 style="color:DodgerBlue">Реализация:</h2>


```
public interface ICreditCard
{
    string CardNumber { get; set; }
    string HolderName { get; set; }
    DateTime ExpiryDate { get; set; }
    decimal Balance { get; set; }
    
    string GetInfo();
    void Pay(decimal amount);
    void CheckBalance();
}

public interface IGoldCreditCard : ICreditCard
{
    int BonusMiles { get; set; }
}

public interface IPremiumCreditCard : ICreditCard
{
    string SupportAssistant { get; set; }
}

public interface ICorporateCreditCard : ICreditCard
{
    string Company { get; set; }
}

public class CreditCard : ICreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public DateTime ExpiryDate { get; set; }
    public decimal Balance { get; set; }

    public CreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal balance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;
    }

    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, Срок: {ExpiryDate:MM/yy}, Баланс: {Balance:F2} руб.";
    }

    public virtual void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }

        if (Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"Оплачено: {amount:F2} руб. Остаток: {Balance:F2} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств! Баланс: {Balance:F2} руб.");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Баланс карты {CardNumber}: {Balance:F2} руб.");
        Console.WriteLine($"Срок действия: {ExpiryDate:MM/yy}");
    }
}

public class GoldCreditCard : CreditCard, IGoldCreditCard
{
    public int BonusMiles { get; set; }

    public GoldCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal balance, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        BonusMiles = bonusMiles;
    }

    public override void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }

        if (Balance >= amount)
        {
            Balance -= amount;
            int miles = (int)(amount / 100);
            BonusMiles += miles;
            Console.WriteLine($"Оплачено: {amount:F2} руб. Начислено миль: {miles}. Всего миль: {BonusMiles}. Остаток: {Balance:F2} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств! Баланс: {Balance:F2} руб.");
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Мили: {BonusMiles}";
    }
}

public class PremiumCreditCard : CreditCard, IPremiumCreditCard
{
    public string SupportAssistant { get; set; }

    public PremiumCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal balance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        SupportAssistant = supportAssistant;
    }

    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"Персональный ассистент: {SupportAssistant}");
        Console.WriteLine("Если у вас проблемы с балансом, обратитесь к ассистенту!");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Ассистент: {SupportAssistant}";
    }
}

public class CorporateCreditCard : CreditCard, ICorporateCreditCard
{
    public string Company { get; set; }

    public CorporateCreditCard(string cardNumber, string holderName, DateTime expiryDate, decimal balance, string company)
        : base(cardNumber, holderName, expiryDate, balance)
    {
        Company = company;
    }

    public override string GetInfo()
    {
        return $"КОРПОРАТИВНАЯ КАРТА\nКомпания: {Company}\n{base.GetInfo()}";
    }

    public override void Pay(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Сумма должна быть больше 0");
            return;
        }

        if (Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"Корпоративный платеж от {Company}: {amount:F2} руб. Остаток: {Balance:F2} руб.");
        }
        else
        {
            Console.WriteLine($"Недостаточно корпоративных средств! Баланс: {Balance:F2} руб.");
            Console.WriteLine($"Уведомление отправлено в финансовый отдел {Company}");
        }
    }
}

// Тестирование
var regularCard = new CreditCard("1212-2121-3443-5445", "Иван Анатольевич", new DateTime(2026, 12, 31), 3200.00m);
var goldCard = new GoldCreditCard("5555-6666-7777-8888", "Пелогея Иванова", new DateTime(2025, 6, 30), 6781.00m, 100);
var premiumCard = new PremiumCreditCard("9173-8888-7647-6568", "Александр Петров", new DateTime(2027, 3, 15), 14234.10m, "Мария Соколова");
var corporateCard = new CorporateCreditCard("7128-3393-2221-3211", "Сергей Смирнов", new DateTime(2024, 9, 30), 71002.00m, "ООО Веники Иваныча ");

Console.WriteLine("=== ИНФОРМАЦИЯ О КАРТАХ ===");
Console.WriteLine(regularCard.GetInfo());
Console.WriteLine();
Console.WriteLine(goldCard.GetInfo());
Console.WriteLine();
Console.WriteLine(premiumCard.GetInfo());
Console.WriteLine();
Console.WriteLine(corporateCard.GetInfo());
Console.WriteLine();

Console.WriteLine("=== ПРОВЕРКА БАЛАНСА ===");
regularCard.CheckBalance();
Console.WriteLine();
goldCard.CheckBalance();
Console.WriteLine();
premiumCard.CheckBalance();
Console.WriteLine();
corporateCard.CheckBalance();
Console.WriteLine();

Console.WriteLine("=== ОПЛАТА ПОКУПОК ===");
Console.WriteLine("Обычная карта:");
regularCard.Pay(1200.00m);
Console.WriteLine();

Console.WriteLine("Золотая карта:");
goldCard.Pay(2500.00m);
Console.WriteLine();

Console.WriteLine("Премиум карта:");
premiumCard.Pay(3000.00m);
Console.WriteLine();

Console.WriteLine("Корпоративная карта:");
corporateCard.Pay(15000.00m);
Console.WriteLine();

Console.WriteLine("=== ПОПЫТКА ОПЛАТЫ СВЕРХ БАЛАНСА ===");
regularCard.Pay(5000.00m);
Console.WriteLine();

Console.WriteLine("=== ПОЛИМОРФИЗМ ===");
ICreditCard[] cards = { regularCard, goldCard, premiumCard, corporateCard };
foreach (var card in cards)
{
    Console.WriteLine($"Вызов Pay(300) для {card.GetType().Name}:");
    card.Pay(300.00m);
    Console.WriteLine();
}

Console.WriteLine("=== ИТОГОВАЯ ИНФОРМАЦИЯ ===");
Console.WriteLine(regularCard.GetInfo());
Console.WriteLine();
Console.WriteLine(goldCard.GetInfo());
Console.WriteLine();
Console.WriteLine(premiumCard.GetInfo());
Console.WriteLine();
Console.WriteLine(corporateCard.GetInfo());
```
