```polyglot-notebook
 
    using System;

class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} выросло на {growth} см. Теперь высота: {Height} см.");
    }
}

class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, string color, int petalCount, float height)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}

class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, int age, string leafType, float height)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowUp(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} постарело на {years} год(а). Теперь ему {Age} лет.");
    }
}

class Program
{
    static void Main(string[] args)
    {
        Flower rose = new Flower("Роза", "Красный", 20, 52f);
        Tree oak = new Tree("Дуб", 12, "Лиственные", 420f);

        Console.WriteLine("Цветение:");
        rose.Bloom();
        oak.Bloom();

        Console.WriteLine("\nРост:");
        rose.Grow(4.2f);
        oak.Grow(12.0f);

        Console.WriteLine("\nДополнительно:");
        rose.Wilt();
        oak.GrowUp(5);

        Console.WriteLine("\nПолиморфизм:");
        Plant[] plants = { rose, oak };
        foreach (Plant plant in plants)
        {
            plant.Bloom();
        }

        Console.WriteLine("\nНажмите клавишу для выхода...");
        Console.ReadKey();
    }
}
```
